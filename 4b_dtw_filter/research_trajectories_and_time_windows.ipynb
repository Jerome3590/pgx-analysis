# DTW Filter Research: Trajectories and Time Windows Analysis

This notebook helps research and classify trajectories, time windows, and codes to identify:
- **Clinical/Procedural**: Useful signals for prediction (keep)
- **Administrative/Post-Event**: Noise that should be filtered (remove)

## Research Objectives

1. **Time Windows**: What intervals between events are meaningful?
   - Protocol-like care: < 7 days (routine follow-ups, refills)
   - Clinical deviations: > 7 days (predictive signals)

2. **Common Trajectories**: Which sequence patterns appear frequently?
   - Standard care pathways (both targets and controls follow)
   - Predictive patterns (deviations from standard care)

3. **Code Classification**: Which codes are clinical vs. administrative?
   - Clinical codes: Diagnoses, procedures, medications (keep)
   - Administrative codes: Billing, scheduling, post-event documentation (filter)

4. **Post-Event Events**: Which events occur after the target event?
   - These are leakage and should be filtered
   - Use target event date as cutoff (matching BupaR logic)
