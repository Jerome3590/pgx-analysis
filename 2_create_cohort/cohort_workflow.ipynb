## C. Create Cohorts

This section creates cohorts with the new time-windowed HCG event logic.

### Cohort Types:

1. **OPIOID_ED** (`opioid_ed`): Uses F11.20 as target event (no time windowing)
2. **POLYPHARMACY** (`non_opioid_ed`): Uses time-windowed HCG events as target

### Time Windows:

For the polypharmacy cohort, multiple time windows are created:
- 7 days
- 14 days (default)
- 21 days
- 30 days
- 45 days

The main `is_target_case` column uses the specified `TIME_WINDOW_DAYS`.
Additional columns (`is_target_case_7d`, `is_target_case_14d`, etc.) are created for multiclass analysis.

```python
# Create a single cohort/age_band/year combination
# Modify these variables to create specific cohorts

COHORT = "non_opioid_ed"  # or "opioid_ed"
AGE_BAND = "65-74"        # Change as needed
EVENT_YEAR = 2016          # Change as needed

print(f"🚀 Creating cohort: {COHORT} / {AGE_BAND} / {EVENT_YEAR}")
print(f"   Started at: {datetime.now()}")
print("=" * 60)

# Build command
cmd = [
    str(PYTHON_BIN),
    str(PROJECT_ROOT / "2_create_cohort" / "0_create_cohort.py"),
    "--age-band", AGE_BAND,
    "--event-year", str(EVENT_YEAR)
]

# Add cohort type
if COHORT == "opioid_ed":
    cmd.extend(["--cohort", "opioid_ed"])
elif COHORT == "non_opioid_ed":
    cmd.extend(["--cohort", "ed_non_opioid"])
    # Add time window for polypharmacy cohort
    cmd.extend(["--time-window-days", str(TIME_WINDOW_DAYS)])

# Add log file
log_file = LOG_DIR / f"create_cohort_{COHORT}_{AGE_BAND}_{EVENT_YEAR}_{datetime.now().strftime('%Y%m%d_%H%M%S')}.log"

print(f"   Command: {' '.join(cmd)}")
print(f"   Log file: {log_file}")
print("=" * 60)

# Run command
with open(log_file, 'w') as f:
    result = subprocess.run(
        cmd,
        cwd=str(PROJECT_ROOT),
        stdout=f,
        stderr=subprocess.STDOUT,
        text=True
    )

if result.returncode == 0:
    print(f"\n✅ Cohort creation completed successfully")
    print(f"   Log: {log_file}")
else:
    print(f"\n❌ Cohort creation failed with return code {result.returncode}")
    print(f"   Check log: {log_file}")
```

```python
# Create all cohorts for all age bands and years
# This will process all combinations sequentially

import time
from concurrent.futures import ThreadPoolExecutor, as_completed

print(f"🚀 Creating all cohorts...")
print(f"   Started at: {datetime.now()}")
print(f"   Cohorts: {COHORTS}")
print(f"   Time Window (polypharmacy): {TIME_WINDOW_DAYS} days")
print("=" * 60)

# Build list of all jobs
jobs = []
for cohort in COHORTS:
    for age_band in AGE_BANDS.get(cohort, []):
        for event_year in EVENT_YEARS:
            jobs.append({
                'cohort': cohort,
                'age_band': age_band,
                'event_year': event_year
            })

print(f"   Total jobs: {len(jobs)}")
print("=" * 60)

# Function to run a single job
def run_cohort_job(job):
    cohort = job['cohort']
    age_band = job['age_band']
    event_year = job['event_year']
    
    job_id = f"{cohort}/{age_band}/{event_year}"
    
    # Build command
    cmd = [
        str(PYTHON_BIN),
        str(PROJECT_ROOT / "2_create_cohort" / "0_create_cohort.py"),
        "--age-band", age_band,
        "--event-year", str(event_year)
    ]
    
    # Add cohort type
    if cohort == "opioid_ed":
        cmd.extend(["--cohort", "opioid_ed"])
    elif cohort == "non_opioid_ed":
        cmd.extend(["--cohort", "ed_non_opioid"])
        # Add time window for polypharmacy cohort
        cmd.extend(["--time-window-days", str(TIME_WINDOW_DAYS)])
    
    # Log file
    log_file = LOG_DIR / f"create_cohort_{cohort}_{age_band}_{event_year}_{datetime.now().strftime('%Y%m%d_%H%M%S')}.log"
    
    try:
        start_time = time.time()
        
        with open(log_file, 'w') as f:
            result = subprocess.run(
                cmd,
                cwd=str(PROJECT_ROOT),
                stdout=f,
                stderr=subprocess.STDOUT,
                text=True,
                timeout=3600  # 1 hour timeout per job
            )
        
        elapsed = time.time() - start_time
        
        if result.returncode == 0:
            return {
                'job_id': job_id,
                'status': 'SUCCESS',
                'elapsed': elapsed,
                'log': log_file
            }
        else:
            return {
                'job_id': job_id,
                'status': 'FAILED',
                'returncode': result.returncode,
                'elapsed': elapsed,
                'log': log_file
            }
    except subprocess.TimeoutExpired:
        return {
            'job_id': job_id,
            'status': 'TIMEOUT',
            'log': log_file
        }
    except Exception as e:
        return {
            'job_id': job_id,
            'status': 'ERROR',
            'error': str(e),
            'log': log_file
        }

# Run jobs sequentially (to avoid resource contention)
# For parallel execution, use ThreadPoolExecutor with max_workers=2
MAX_WORKERS = 1  # Set to 2 for parallel execution (be careful with S3 rate limits)

results = []
if MAX_WORKERS == 1:
    # Sequential execution
    for i, job in enumerate(jobs, 1):
        print(f"\n[{i}/{len(jobs)}] Processing: {job['cohort']}/{job['age_band']}/{job['event_year']}")
        result = run_cohort_job(job)
        results.append(result)
        print(f"   Status: {result['status']}")
        if 'elapsed' in result:
            print(f"   Elapsed: {result['elapsed']:.1f}s")
else:
    # Parallel execution
    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
        future_to_job = {executor.submit(run_cohort_job, job): job for job in jobs}
        
        for i, future in enumerate(as_completed(future_to_job), 1):
            job = future_to_job[future]
            try:
                result = future.result()
                results.append(result)
                print(f"\n[{i}/{len(jobs)}] {result['job_id']}: {result['status']}")
            except Exception as e:
                print(f"\n[{i}/{len(jobs)}] {job['cohort']}/{job['age_band']}/{job['event_year']}: ERROR - {e}")

# Summary
print("\n" + "=" * 60)
print("📊 SUMMARY")
print("=" * 60)

success_count = sum(1 for r in results if r['status'] == 'SUCCESS')
failed_count = sum(1 for r in results if r['status'] in ['FAILED', 'ERROR', 'TIMEOUT'])

print(f"✅ Successful: {success_count}/{len(jobs)}")
print(f"❌ Failed: {failed_count}/{len(jobs)}")

if failed_count > 0:
    print("\nFailed jobs:")
    for r in results:
        if r['status'] in ['FAILED', 'ERROR', 'TIMEOUT']:
            print(f"   {r['job_id']}: {r['status']} (log: {r.get('log', 'N/A')})")
```

## D. Verify Cohort Creation

This section verifies that cohorts were created successfully by checking S3 and local paths.

```python
# Verify cohorts exist in S3
s3_client = boto3.client('s3')

print("📊 Verifying cohorts in S3...")
print("=" * 60)

cohort_mapping = {
    "opioid_ed": "opioid_ed",
    "non_opioid_ed": "ed_non_opioid"
}

all_found = True
for cohort in COHORTS:
    s3_cohort_name = cohort_mapping.get(cohort, cohort)
    
    for age_band in AGE_BANDS.get(cohort, []):
        for event_year in EVENT_YEARS:
            s3_key = f"gold/cohorts/cohort_name={s3_cohort_name}/age_band={age_band}/event_year={event_year}/cohort.parquet"
            
            try:
                s3_client.head_object(Bucket=S3_BUCKET, Key=s3_key)
                # Get file size
                response = s3_client.head_object(Bucket=S3_BUCKET, Key=s3_key)
                size_mb = response['ContentLength'] / (1024 * 1024)
                print(f"✅ {cohort}/{age_band}/{event_year}: {size_mb:.2f} MB")
            except s3_client.exceptions.ClientError as e:
                if e.response['Error']['Code'] == '404':
                    print(f"❌ {cohort}/{age_band}/{event_year}: NOT FOUND")
                    all_found = False
                else:
                    print(f"⚠️  {cohort}/{age_band}/{event_year}: ERROR - {e}")
                    all_found = False

print("=" * 60)
if all_found:
    print("✅ All cohorts verified in S3")
else:
    print("❌ Some cohorts are missing in S3")
```

```python
# Verify multiclass target columns exist for polypharmacy cohort
import duckdb

print("📊 Verifying multiclass target columns for polypharmacy cohort...")
print("=" * 60)

# Connect to DuckDB and enable S3
con = duckdb.connect()
con.execute("INSTALL httpfs; LOAD httpfs;")
con.execute("INSTALL aws; LOAD aws;")
con.execute("CALL load_aws_credentials();")

# Check one polypharmacy cohort file
test_cohort = "non_opioid_ed"
test_age_band = AGE_BANDS[test_cohort][0]  # First age band
test_year = EVENT_YEARS[0]  # First year

s3_path = f"s3://{S3_BUCKET}/gold/cohorts/cohort_name=ed_non_opioid/age_band={test_age_band}/event_year={test_year}/cohort.parquet"

try:
    # Get schema
    schema_query = f"DESCRIBE SELECT * FROM read_parquet('{s3_path}') LIMIT 0"
    schema = con.execute(schema_query).fetchdf()
    
    print(f"\nColumns in {test_cohort}/{test_age_band}/{test_year}:")
    
    # Check for multiclass columns
    expected_columns = [
        'is_target_case',
        'is_target_case_7d',
        'is_target_case_14d',
        'is_target_case_21d',
        'is_target_case_30d',
        'is_target_case_45d'
    ]
    
    found_columns = []
    missing_columns = []
    
    for col in expected_columns:
        if col in schema['column_name'].values:
            found_columns.append(col)
            print(f"   ✅ {col}")
        else:
            missing_columns.append(col)
            print(f"   ❌ {col} (missing)")
    
    if len(missing_columns) == 0:
        print("\n✅ All multiclass target columns found")
    else:
        print(f"\n❌ Missing columns: {missing_columns}")
        
except Exception as e:
    print(f"❌ Error checking schema: {e}")

con.close()
```
